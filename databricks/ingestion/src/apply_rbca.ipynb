```python
# Notebook: apply_rbca
# Purpose: Applies role-based access control (RBAC) grants on the lh_nautical.bronze schema
#          and on the LGPD masking function for the stakeholders, engenharia_dados and time_negocios groups.

from databricks.sdk.runtime import spark
from grupo10.utils.logger import get_logger

logger = get_logger("apply_rbac")

#Grants access to the 'stakeholders' group
spark.sql("GRANT SELECT ON SCHEMA lh_nautical.bronze TO `stakeholders`;")
logger.info("Granted SELECT to stakeholders")

#Grants access to the 'engenharia_dados' group
spark.sql("GRANT SELECT, MODIFY ON SCHEMA lh_nautical.bronze TO `engenharia_dados`;")
logger.info("Granted SELECT, MODIFY to engenharia_dados")

#Grants access to the 'time_negocios' group
spark.sql("GRANT SELECT ON SCHEMA lh_nautical.bronze TO `time_negocios`;")
logger.info("Granted SELECT to time_negocios")

#Grants permission to query masked tables (EXECUTE on the masking function)
spark.sql("GRANT EXECUTE ON FUNCTION lh_nautical.bronze.regra_mascaramento TO `stakeholders`;")
spark.sql("GRANT EXECUTE ON FUNCTION lh_nautical.bronze.regra_mascaramento TO `engenharia_dados`;")
spark.sql("GRANT EXECUTE ON FUNCTION lh_nautical.bronze.regra_mascaramento TO `time_negocios`;")

logger.info("Granted EXECUTE on masking function to all groups")
```
